Day 143 using Faster-Whisper Approach 1

```python
# 1. Install faster-whisper
!pip install -q faster-whisper

from google.colab import files
from faster_whisper import WhisperModel
import os

# 2. Select your uploaded MP3 file (or upload it again)
print("Click 'Choose Files' to select your MP3 file:")
uploaded = files.upload()

if uploaded:
    audio_file = list(uploaded.keys())[0]
    print(f"\nProcessing audio file: {audio_file}")

    # 3. Load Medium Model (Lightweight RAM footprint, highly accurate for inquiry terms)
    print("\nLoading Memory-Optimized Whisper Model on GPU...")
    model = WhisperModel("medium", device="cuda", compute_type="float16")

    # 4. Transcribe with streaming output
    print("\nTranscribing... (This will take ~3-5 minutes)")
    segments, info = model.transcribe(
        audio_file,
        beam_size=2,                # Lower beam size dramatically cuts RAM/VRAM usage
        language="en",
        condition_on_previous_text=False  # Prevents memory accumulation loops
    )

    print(f"\nDetected Language: '{info.language}' ({info.language_probability*100:.1f}% confidence)\n")

    # 5. Save transcript line-by-line
    output_filename = "FW_Vennells_Day143.txt"
    with open(output_filename, "w", encoding="utf-8") as f:
        for segment in segments:
            line = f"[{segment.start:.2f}s -> {segment.end:.2f}s] {segment.text}\n"
            f.write(line)
            print(line, end="")

    print(f"\n\nSUCCESS! Saved as {output_filename}")

    # 6. Download transcript file to laptop
    files.download(output_filename)
```

```python
import assemblyai as aai

# 1. Set your API key
from google.colab import userdata
aai.settings.api_key = userdata.get('assemblyaikey')
# 2. Configure transcription parameters
# Enable custom vocabulary boosting for Post Office legal jargon
config = aai.TranscriptionConfig(
    word_boost=["Vennells", "Fujitsu", "Second Sight", "Horizon", "Crichton", "Jenkins"],
    boost_param="high",
    speaker_labels=True  # Enables speaker identification (Diarization)
)

# 3. Specify the audio file (90-second snippet or full MP3 file)
audio_file_path = "Y2Mate.mp3"

# 4. Initialize the transcriber and process audio
transcriber = aai.Transcriber()
print("Transcribing via AssemblyAI API...")
transcript = transcriber.transcribe(audio_file_path, config=config)

# 5. Check status and output result
if transcript.status == aai.TranscriptStatus.error:
    print(f"Error: {transcript.error}")
else:
    # Save clean transcript text
    with open("API_Vennells_Day143.txt", "w", encoding="utf-8") as f:
        f.write(transcript.text)

    print("Success! Transcript saved as API_Vennells_Day143.txt")
```

```python
# Alan Bates - Day 117 AM (9 April 2024) - Post Office Horizon IT Inquiry
# AssemblyAI RAW transcript only
!pip install -q assemblyai
import assemblyai as aai
from google.colab import files, userdata

aai.settings.api_key = userdata.get('assemblyaikey')

# --- UPLOAD YOUR MP3 ---
print("Upload the Bates Day 117 AM mp3:")
uploaded = files.upload()
audio_file_path = list(uploaded.keys())[0]

config = aai.TranscriptionConfig(
    word_boost=["Vennells", "Fujitsu", "Second Sight", "Horizon", "Bates",
                "Beer", "subpostmaster", "Justice for Subpostmasters"],
    boost_param="high",
    speaker_labels=False
)

transcriber = aai.Transcriber()
print("Transcribing via AssemblyAI... this runs in the background, feel free to wait.")
transcript = transcriber.transcribe(audio_file_path, config=config)

if transcript.status == aai.TranscriptStatus.error:
    print(f"Error: {transcript.error}")
else:
    with open("API_Bates_Day117_AM.txt", "w", encoding="utf-8") as f:
        f.write(transcript.text)
    print("Transcript saved: API_Bates_Day117_AM.txt")
    files.download("API_Bates_Day117_AM.txt")
```
